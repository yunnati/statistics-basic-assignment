Q. 1: What is the difference between descriptive statistics and inferential statistics? Explain with examples.
- Descriptive statistics focuses on collecting, organizing, summarizing, and presenting the data that has already been collected.
- It helps us understand the main characteristics of data through measures such as mean, median, mode, standard deviation, and graphs such as histograms and bar charts.
- It describes the available dataset rather than making predictions about data outside it.
- Example: If a teacher finds the average marks of 40 students in a class, it is an example of descriptive statistics because only those 40 students' marks are being summarized.
- Inferential statistics uses information obtained from a sample to draw conclusions, make predictions, or make decisions about a larger population.
- Common methods include hypothesis testing, confidence intervals, and regression analysis.
- Example: If 500 people from a city are surveyed about their preferred mobile brand and the result is used to estimate the preference of the entire city, it is inferential statistics.

Q. 2: What is sampling in statistics? Explain the differences between random and stratified sampling.
- Sampling means choosing a smaller set of individuals from a larger population for analysis. The selected group is called a sample.
- It is useful because studying the complete population can require a large amount of time, money, and effort.
- Random Sampling: In this method, each member of the population has an equal opportunity to be selected. The selection is made randomly, such as choosing names from a box.
- Random sampling is simple, but when a population contains different small groups, some groups may not be properly represented.
- Stratified Sampling: Here, the population is separated into groups called strata according to a common characteristic such as age, gender, or income. Samples are then selected from each group.
This method helps make sure that the important subgroups are represented.
- Example: Suppose a college consists of 60% boys and 40% girls. Stratified sampling can select students in the same 60:40 proportion. Random sampling would select students directly from the complete student list without first separating them into groups.

Q. 3: Define mean, median, and mode. Explain why these measures of central tendency are important.
- Mean: Mean represents the average value of a dataset. It is obtained by adding all observations and dividing their sum by the number of observations.
- Median: Median is the central value after arranging the observations in order. When there are an even number of observations, the two central values are averaged.
- Mode: Mode is the observation that appears most often in a dataset. A dataset may have one mode, multiple modes, or sometimes no mode.
- Importance: Mean, median, and mode provide a simple way of representing a dataset using a central or typical value.
- Mean is useful for finding the overall average, while median is useful when extreme values or outliers may affect the average.
- Mode is helpful for finding the most frequently occurring value and can also be useful with categorical data.

Q. 4: Explain skewness and kurtosis. What does a positive skew imply about the data?
- Skewness describes how much a distribution is asymmetric around its central value. It indicates whether the distribution has more spread toward one side.
- Positive Skew: In a positively skewed distribution, the right-side tail is longer and most observations are concentrated toward the lower values. Generally, the mean becomes greater than the median because unusually large observations pull the mean upward.
- Negative Skew: In a negatively skewed distribution, the left-side tail is longer and most observations are located toward the higher values. In this situation, the mean is generally smaller than the median.
- Kurtosis describes the heaviness of the tails of a distribution compared with a normal distribution. It provides an indication of the possibility of extreme observations.
- High kurtosis, called leptokurtic distribution, has heavier tails and can contain more extreme values. Low kurtosis, called platykurtic distribution, has lighter tails.
- Therefore, positive skew indicates a longer right tail, with most observations toward the lower side and a few unusually large observations.
- Example: Income distribution is commonly positively skewed because many people may have moderate incomes while a small number of people have extremely high incomes.

Q. 6: Compute the covariance and correlation coefficient between the following two
datasets provided as lists in Python:
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]


```python
'''
import numpy as np
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]
covariance = np.cov(list_x, list_y, ddof=1)[0][1]
correlation = np.corrcoef(list_x, list_y)[0][1]
print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)
'''

'''Output:
Covariance: 275.0
Correlation Coefficient: 0.9959'''
'''Interpretation: The covariance of 275 is positive, meaning list_x and list_y increase together. The
correlation coefficient of about 0.996 is very close to +1, which shows an extremely strong positive linear
relationship between the two datasets.'''
```

Q. 7: Write a Python script to draw a boxplot for the following numeric list and identify
its outliers. Explain the result:
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]


```python
'''
import matplotlib.pyplot as plt
import numpy as np
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = [x for x in data if x < lower_bound or x > upper_bound]
print("Q1:", Q1, "Q3:", Q3, "IQR:", IQR)
print("Lower Bound:", lower_bound, "Upper Bound:", upper_bound)
print("Outliers:", outliers)
plt.boxplot(data, vert=True)
plt.title("Boxplot of Data")
plt.ylabel("Values")
plt.savefig("boxplot.png")'''

'''Output:
Q1: 17.25 Q3: 23.25 IQR: 6.0
Lower Bound: 8.25 Upper Bound: 32.25
Outliers: [35]'''
'''Explanation: The box in the boxplot represents the interquartile range (IQR) — the middle 50% of the
data, from Q1 (17.25) to Q3 (23.25). The line inside the box is the median. The 'whiskers' extend to the
smallest and largest values within 1.5 times the IQR from the box. Any point beyond the whiskers is
considered an outlier. Here, the value 35 lies above the upper bound of 32.25, so it is flagged as an
outlier — an unusually high value compared to the rest of the data.
```

Q. 8: You are working as a data analyst in an e-commerce company. The marketing team
wants to know if there is a relationship between advertising spend and daily sales. Explain how you would use covariance and correlation to explore this relationship., Write Python code to compute the correlation between the two lists:
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]
- Covariance tells us the direction of the relationship between advertising spend and daily sales, whether they move together (positive covariance) or in opposite directions (negative covariance).
- However, covariance's value depends on the units used, so it's hard to judge the strength of the relationship from it alone.
- Correlation (specifically the Pearson correlation coefficient) standardizes this relationship into a
value between -1 and +1, which makes it much easier to interpret. A value close to +1 means
advertising spend and sales increase together strongly; close to -1 would mean they move in
opposite directions; close to 0 would mean little to no linear relationship
- This helps the marketing
team judge how strongly advertising spend is driving sales.

```python
'''
import numpy as np
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]
covariance = np.cov(advertising_spend, daily_sales, ddof=1)[0][1]
correlation = np.corrcoef(advertising_spend, daily_sales)[0][1]
print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)'''

'''Output:
Covariance: 84875.0
Correlation Coefficient: 0.9936'''
'''Interpretation: The correlation coefficient of about 0.99 shows a very strong positive relationship — as
advertising spend increases, daily sales increase almost proportionally. This suggests advertising is likely
a strong driver of sales for this company.'''
```

Q. 9: Your team has collected customer satisfaction survey data on a scale of 1-10 and
wants to understand its distribution before launching a new product.
● Explain which summary statistics and visualizations (e.g. mean, standard deviation, histogram)
you'd use.
● Write Python code to create a histogram using Matplotlib for the survey data:
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]
- To understand the distribution of customer satisfaction scores, I would use the mean (to know the
average satisfaction level), the median (to check the typical score without being affected by extreme
ratings), and the standard deviation (to see how spread out or consistent the ratings are , a low
standard deviation means most customers agree, while a high one means opinions vary a lot).
- For visualization, a histogram is very useful here because it shows how frequently each score (or
range of scores) occurs, making it easy to see whether customers are mostly satisfied, mostly
dissatisfied, or split. A boxplot could also help spot any unusual (outlier) ratings.

```python
'''
import matplotlib.pyplot as plt
import statistics
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]
mean_val = statistics.mean(survey_scores)
median_val = statistics.median(survey_scores)
stdev_val = statistics.stdev(survey_scores)
print("Mean:", mean_val)
print("Median:", median_val)
print("Standard Deviation:", stdev_val)
plt.hist(survey_scores, bins=range(1,12), edgecolor='black', align='left')
plt.title("Histogram of Customer Satisfaction Scores")
plt.xlabel("Satisfaction Score (1-10)")
plt.ylabel("Frequency")
plt.savefig("histogram.png")'''

'''Output:
Mean: 7.33
Median: 7
Standard Deviation: 1.63'''
'''Interpretation: The mean satisfaction score is about 7.33 out of 10, with a median of 7, showing that
most customers rate the product moderately-to-highly. The standard deviation of about 1.63 indicates
the ratings are fairly close together, meaning customer opinions are fairly consistent (not wildly different
from one another). The histogram confirms this — most scores cluster between 6 and 9, with very few
very low or very high ratings.'''
```
